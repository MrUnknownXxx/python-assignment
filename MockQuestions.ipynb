# INTERVIEW QUESTIONS
1-What is python 

Python is a high-level, general-purpose programming language known for its simple, readable syntax that resembles everyday English. Created by Guido van Rossum and first released in 1991

2- What is high level and low level languuage ?

high-level languages use human-friendly syntax with high abstraction, while low-level languages stay close to hardware instructions and memory

3- Tell me the areas where python is being used?

web development, data science, artificial intelligence, automation, and software engineering

4-What is interpreted and compiled language?

Compiled languages translate your entire source code into machine code before the program runs, while interpreted languages read, translate, and execute your code line by line while the program runs

5-What is statically typed and dynamically typed 5-language




A statically typed language checks variable types before the program runs (at compile time), while a dynamically typed language checks variable types while the program is running (at runtime)

6-What is weekly typed and strongly typed language?

Strongly typed languages strictly enforce data type rules and avoid silent type conversions, while weakly typed languages allow more flexible or implicit type conversions during operations

7-What is.PYC file (Byte code) when it is created?

A .pyc file is a compiled bytecode file that Python automatically generates from your .py source code. It acts as a cache so that Python does not have to recompile the same code every time it runs, speeding up program startup

8-What is PVM?

Parallel Virtual Machine in distributed computing, Python Virtual Machine in programming, Price-Volume-Mix in finance, Product Value Management in business, or Player versus Monster in gaming

9-How Python internally works?


hybrid process that combines both compilation and interpretation. When you run a Python script (specifically in standard CPython), your human-readable .py file goes through an internal pipeline before your CPU executes it.

10- What is PEP 8?

PEP 8 (Python Enhancement Proposal 8) is the official style guide for Python code, created in 2001 by Guido van Rossum, Barry Warsaw, and Alyssa Coghlan

11-What is PIP what is the use of PIP?

PIP most commonly stands for Package Installer for Python (or "Pip Installs Packages"), the standard command-line package manager used to install and manage external Python libraries
